```python
import csv
import gc
import re
from datasets import load_dataset

SPLITS = [
    "asm_Beng",
    "ben_Beng",
    "brx_Deva",
    "doi_Deva",
    "gom_Deva",
    "guj_Gujr",
    "hin_Deva",
    "kan_Knda",
    "kas_Arab",
    "mai_Deva",
    "mal_Mlym",
    "mar_Deva",
    "mni_Mtei",
    "npi_Deva",
    "ory_Orya",
    "pan_Guru",
    "san_Deva",
    "snd_Deva",
    "tam_Taml",
    "tel_Telu",
    "urd_Arab",
    "khasi",
    "santhali",
]

output_file = "indic_corpus_subset.csv"
SENTENCE_SPLITTER = re.compile(r"[।\.\?\!\n]+")

with open(output_file, mode="w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["text", "label"])

    for split_name in SPLITS:
        print(f"Extracting {split_name}...", end=" ", flush=True)
        count = 0

        try:
            # Stream directly with smaller buffer batching
            ds = load_dataset(
                "ai4bharat/IndicCorpV2",
                split=split_name,
                streaming=True,
                trust_remote_code=True,
            )

            for row in ds:
                text = row.get("text", "")
                if not text:
                    continue

                for sent in SENTENCE_SPLITTER.split(text):
                    clean_sent = sent.strip()
                    if len(clean_sent) > 15:
                        writer.writerow([clean_sent, split_name])
                        count += 1

                    if count >= 1000:
                        break

                if count >= 1000:
                    break

            print(f"Done ({count} sentences)")

        except Exception as e:
            print(f"Skipped due to network error: {e}")

        del ds
        gc.collect()

print(f"\nSaved dataset to '{output_file}'")
```

```python
%pip install scipy

import math
import numpy as np
import pandas as pd
from scipy.sparse import csr_matrix

# Load dataset generated from previous block
df = pd.read_csv("indic_corpus_subset.csv")

# Map language string labels to integer IDs (0-22)
labels = sorted(df["label"].unique())
label2id = {lbl: idx for idx, lbl in enumerate(labels)}
id2label = {idx: lbl for idx, lbl in enumerate(labels)}
df["label_id"] = df["label"].map(label2id)


# Stratified Sampling: 80% Train, 10% Validation, 10% Test
def stratified_split(df, label_col="label_id", train_ratio=0.8, val_ratio=0.1):
    train_dfs, val_dfs, test_dfs = [], [], []

    for label, group in df.groupby(label_col):
        group = group.sample(frac=1.0, random_state=42).reset_index(drop=True)
        n = len(group)
        n_train = int(n * train_ratio)
        n_val = int(n * val_ratio)

        train_dfs.append(group.iloc[:n_train])
        val_dfs.append(group.iloc[n_train : n_train + n_val])
        test_dfs.append(group.iloc[n_train + n_val :])

    train_df = (
        pd.concat(train_dfs)
        .sample(frac=1.0, random_state=42)
        .reset_index(drop=True)
    )
    val_df = (
        pd.concat(val_dfs)
        .sample(frac=1.0, random_state=42)
        .reset_index(drop=True)
    )
    test_df = (
        pd.concat(test_dfs)
        .sample(frac=1.0, random_state=42)
        .reset_index(drop=True)
    )

    return train_df, val_df, test_df


train_df, val_df, test_df = stratified_split(df)
print(
    f"Splits Created -> Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}"
)
```

```python
class CustomTFIDFVectorizer:

    def __init__(self, max_features=25000):
        self.max_features = max_features
        self.vocab = {}
        self.idf = {}

    def _extract_ngrams(self, text):
        if not isinstance(text, str) or not text.strip():
            return []

        words = text.split()
        ngrams = []

        # i. Word Unigrams and Word Bigrams
        ngrams.extend(words)
        for i in range(len(words) - 1):
            ngrams.append(f"{words[i]}_{words[i+1]}")

        # ii. Character 2-grams, 3-grams, and 4-grams
        for n in [2, 3, 4]:
            for i in range(len(text) - n + 1):
                ngrams.append(text[i : i + n])

        return ngrams

    def fit(self, corpus):
        df_counts = {}
        total_docs = len(corpus)

        for doc in corpus:
            for ngram in set(self._extract_ngrams(doc)):
                df_counts[ngram] = df_counts.get(ngram, 0) + 1

        sorted_ngrams = sorted(
            df_counts.items(), key=lambda x: x[1], reverse=True
        )[: self.max_features]

        self.vocab = {
            ngram: idx for idx, (ngram, _) in enumerate(sorted_ngrams)
        }

        for ngram, df_val in sorted_ngrams:
            self.idf[ngram] = math.log((total_docs + 1) / (df_val + 1)) + 1.0

        print(f"Vocabulary size: {len(self.vocab)} n-grams")

    def transform(self, corpus):
        rows, cols, data = [], [], []

        for row_idx, doc in enumerate(corpus):
            ngrams = self._extract_ngrams(doc)
            if not ngrams:
                continue

            tf_counts = {}
            for ngram in ngrams:
                if ngram in self.vocab:
                    tf_counts[ngram] = tf_counts.get(ngram, 0) + 1

            total_terms = len(ngrams)
            for ngram, count in tf_counts.items():
                col_idx = self.vocab[ngram]
                tf = count / total_terms
                tfidf_val = tf * self.idf[ngram]

                rows.append(row_idx)
                cols.append(col_idx)
                data.append(tfidf_val)

        return csr_matrix(
            (data, (rows, cols)),
            shape=(len(corpus), len(self.vocab)),
            dtype=np.float32,
        )

    def fit_transform(self, corpus):
        self.fit(corpus)
        return self.transform(corpus)
```

```python
print("Extracting features for Train, Validation, and Test sets...")
vectorizer = CustomTFIDFVectorizer(max_features=25000)

X_train = vectorizer.fit_transform(train_df["text"])
X_val = vectorizer.transform(val_df["text"])
X_test = vectorizer.transform(test_df["text"])

y_train = train_df["label_id"].values
y_val = val_df["label_id"].values
y_test = test_df["label_id"].values
```

```python
def custom_macro_f1(y_true, y_pred, eps=1e-9):
    unique_classes = np.unique(y_true)
    f1_scores = []

    for c in unique_classes:
        tp = np.sum((y_true == c) & (y_pred == c))
        fp = np.sum((y_true != c) & (y_pred == c))
        fn = np.sum((y_true == c) & (y_pred != c))

        precision = tp / (tp + fp + eps)
        recall = tp / (tp + fn + eps)
        f1 = (2 * precision * recall) / (precision + recall + eps)

        f1_scores.append(f1)

    return float(np.mean(f1_scores))


class CustomLogisticRegression:

    def __init__(self, num_classes, learning_rate=0.5, epochs=15, batch_size=256):
        self.num_classes = num_classes
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        self.W = None
        self.b = None

    def _softmax(self, Z):
        exp_Z = np.exp(Z - np.max(Z, axis=1, keepdims=True))
        return exp_Z / np.sum(exp_Z, axis=1, keepdims=True)

    def fit(self, X, y, X_val=None, y_val=None):
        num_samples, num_features = X.shape

        self.W = np.zeros((num_features, self.num_classes), dtype=np.float32)
        self.b = np.zeros((1, self.num_classes), dtype=np.float32)

        Y_onehot = np.zeros((num_samples, self.num_classes), dtype=np.float32)
        Y_onehot[np.arange(num_samples), y] = 1.0

        for epoch in range(1, self.epochs + 1):
            indices = np.arange(num_samples)
            np.random.shuffle(indices)

            for start in range(0, num_samples, self.batch_size):
                end = min(start + self.batch_size, num_samples)
                batch_idx = indices[start:end]

                X_batch = X[batch_idx]
                Y_batch = Y_onehot[batch_idx]
                curr_batch_size = end - start

                # Forward pass
                Z = X_batch.dot(self.W) + self.b
                P = self._softmax(Z)

                # Gradient step
                error = P - Y_batch
                dW = (X_batch.T.dot(error)) / curr_batch_size
                db = np.sum(error, axis=0, keepdims=True) / curr_batch_size

                self.W -= self.lr * dW
                self.b -= self.lr * db

            # Log validation metric
            train_preds = self.predict(X)
            train_f1 = custom_macro_f1(y, train_preds)

            if X_val is not None and y_val is not None:
                val_preds = self.predict(X_val)
                val_f1 = custom_macro_f1(y_val, val_preds)
                print(
                    f"Epoch {epoch:02d}/{self.epochs:02d} | Train F1: {train_f1:.4f} | Val F1: {val_f1:.4f}"
                )

    def predict(self, X):
        Z = X.dot(self.W) + self.b
        P = self._softmax(Z)
        return np.argmax(P, axis=1)
```

```python
# Train model on custom TF-IDF features
model = CustomLogisticRegression(
    num_classes=len(labels), learning_rate=0.5, epochs=15, batch_size=256
)
model.fit(X_train, y_train, X_val, y_val)

# Evaluate on Test Set
test_preds = model.predict(X_test)
test_f1 = custom_macro_f1(y_test, test_preds)

print(f"\nFinal Test Set Macro-F1 Score: {test_f1:.4f}")
```

```python
def predict_language(sentence):
    vec = vectorizer.transform([sentence])
    
    Z = vec.dot(model.W) + model.b
    probabilities = model._softmax(Z)[0]
    
    pred_id = np.argmax(probabilities)
    confidence = probabilities[pred_id] * 100
    
    predicted_lang = id2label[pred_id]
    
    return predicted_lang, confidence

random_sentences = [
    "यह एक बहुत अच्छा दिन है।",
    "हे एक सुंदर शहर आहे.",
    "ನಾನು ಶಾಲೆಗೆ ಹೋಗುತ್ತೇನೆ.",
    "আমি তোমাকে ভালোবাসি।"
]

print("--- Random Sentence Predictions ---")
for text in random_sentences:
    lang, conf = predict_language(text)
    print(f"Text: {text}")
    print(f"Prediction: {lang} | Confidence: {conf:.2f}%\n")
```

```python
def predict_all_languages(sentence):
    vec = vectorizer.transform([sentence])
    
    Z = vec.dot(model.W) + model.b
    probabilities = model._softmax(Z)[0]
    
    confidences = {}
    for i, prob in enumerate(probabilities):
        lang = id2label[i]
        confidences[lang] = prob * 100
        
    sorted_confidences = sorted(confidences.items(), key=lambda item: item[1], reverse=True)
    
    return sorted_confidences

random_sentences = [
    "भारत के इतिहास में कई महान नेताओं ने देश की स्वतंत्रता के लिए अपना जीवन समर्पित कर दिया, जिसके परिणामस्वरूप आज हम एक स्वतंत्र और लोकतांत्रिक देश में रह रहे हैं।",
    # Marathi
    "महाराष्ट्रातील गडकिल्ले हे केवळ दगड-मातीचे ढिगारे नसून ते छत्रपती शिवाजी महाराजांच्या आणि मावळ्यांच्या शौर्याची आणि पराक्रमाची साक्ष देतात.",
    # Kannada
    "ಭಾರತದ ಬಾಹ್ಯಾಕಾಶ ಸಂಶೋಧನಾ ಸಂಸ್ಥೆಯು ಚಂದ್ರಯಾನ ಮತ್ತು ಮಂಗಳಯಾನದಂತಹ ಪ್ರಮುಖ ಯೋಜನೆಗಳನ್ನು ಯಶಸ್ವಿಯಾಗಿ ಪೂರ್ಣಗೊಳಿಸುವ ಮೂಲಕ ವಿಶ್ವಮಟ್ಟದಲ್ಲಿ ತನ್ನದೇ ಆದ ಛಾಪು ಮೂಡಿಸಿದೆ.",
    # Bengali
    "রবীন্দ্রনাথ ঠাকুর শুধু একজন কবিই ছিলেন না, তিনি ছিলেন একাধারে ঔপন্যাসিক, সঙ্গীতস্রষ্টা, নাট্যকার, চিত্রকর, ছোটগল্পকার, প্রাবন্ধিক, অভিনেতা, কণ্ঠশিল্পী ও দার্শনিক।"

]

for text in random_sentences:
    print(f"Text: {text}")
    predictions = predict_all_languages(text)
    
    for lang, conf in predictions:
        print(f"{lang}: {conf:.2f}%")
    print("-" * 40)
```

```python
# --- Lab 2: Extended TF-IDF Vectorizer supporting all 6 combinations ---

class Lab2TFIDFVectorizer(CustomTFIDFVectorizer):
    def transform_with_config(self, corpus, tf_mode='raw', idf_mode='unnormalized'):
        rows, cols, data = [], [], []
        max_idf = max(self.idf.values()) if self.idf else 1.0
        
        for row_idx, doc in enumerate(corpus):
            ngrams = self._extract_ngrams(doc)
            if not ngrams:
                continue
            
            tf_counts = {}
            for ngram in ngrams:
                if ngram in self.vocab:
                    tf_counts[ngram] = tf_counts.get(ngram, 0) + 1
            
            if not tf_counts:
                continue
                
            total_terms = len(ngrams)
            max_term_freq = max(tf_counts.values()) if tf_counts else 1
            
            for ngram, count in tf_counts.items():
                col_idx = self.vocab[ngram]
                
                # 1. TF Normalization modes
                if tf_mode == 'raw':
                    tf = float(count)
                elif tf_mode == 'total_terms':
                    tf = count / total_terms
                elif tf_mode == 'max_freq':
                    tf = count / max_term_freq
                else:
                    tf = float(count)
                
                # 2. IDF Normalization modes
                raw_idf = self.idf[ngram]
                if idf_mode == 'normalized':
                    idf_val = raw_idf / max_idf
                else:
                    idf_val = raw_idf
                
                tfidf_val = tf * idf_val
                
                rows.append(row_idx)
                cols.append(col_idx)
                data.append(tfidf_val)
                
        return csr_matrix((data, (rows, cols)), shape=(len(corpus), len(self.vocab)), dtype=np.float32)

print("Lab2TFIDFVectorizer initialized successfully.")
```

```python
# --- Train and Evaluate All 6 TF-IDF Combinations ---

tf_modes = ['raw', 'total_terms', 'max_freq']
idf_modes = ['unnormalized', 'normalized']

results = []

# Re-use vocabulary and IDF dictionary from Lab 1 vectorizer
lab2_vectorizer = Lab2TFIDFVectorizer(max_features=25000)
lab2_vectorizer.vocab = vectorizer.vocab
lab2_vectorizer.idf = vectorizer.idf

for tf_m in tf_modes:
    for idf_m in idf_modes:
        print(f"\n--- Training with TF: {tf_m} | IDF: {idf_m} ---")
        
        X_train_comb = lab2_vectorizer.transform_with_config(train_df["text"], tf_mode=tf_m, idf_mode=idf_m)
        X_test_comb = lab2_vectorizer.transform_with_config(test_df["text"], tf_mode=tf_m, idf_mode=idf_m)
        
        comb_model = CustomLogisticRegression(
            num_classes=len(labels), learning_rate=0.5, epochs=10, batch_size=256
        )
        comb_model.fit(X_train_comb, y_train)
        
        preds = comb_model.predict(X_test_comb)
        f1 = custom_macro_f1(y_test, preds)
        
        print(f"Test Macro-F1 ({tf_m}, {idf_m}): {f1:.4f}")
        results.append({
            "TF Normalization": tf_m,
            "IDF Normalization": idf_m,
            "Test Macro-F1": f1
        })

results_df = pd.DataFrame(results)
print("\n=== Performance Comparison of All 6 Combinations ===")
print(results_df.to_markdown(index=False))
```
